```python
#https://chiavinik.github.io/fancygeneration/$0
import pandas as pd
# vamos chamar as tabelas que queremos dar merge (mesclar)
catalago_produtos = '/content/catalogo_produtos.csv'
perfil_clientes = '//content/perfil_clientes.csv'
vendas_pedidos = '/content/vendas_pedidos.csv'

df_catalago_produtos = pd.read_csv(catalago_produtos)
df_perfil_clientes = pd.read_csv(perfil_clientes)
df_vendas_pedidos = pd.read_csv(vendas_pedidos)

#merge 3 tabelas
# primeiro fiz merge de tabela catalago_produtos com vendas_pedidos
df_catalago_produtos_vendas_pedidos = pd.merge(df_catalago_produtos, df_vendas_pedidos, left_on='sku_produto', right_on = 'sku_produto', how = 'inner')
#df_nome_tabela = pd.merge(tabela da esquerda, tabela da direita, left_on = 'nome do campo da tabela da esquerda, right_on= 'nome do campo da outra tabela)
display(df_catalago_produtos_vendas_pedidos)
```

```python
#merge 3 tabelas
# segundo fiz merge de tabela cdf_catalago_produtos_vendas_pedidos com perfil_clientes
df_catalago_produtos_vendas_pedidos_perfil_clientes = pd.merge(df_catalago_produtos_vendas_pedidos, df_perfil_clientes, left_on='id_cliente', right_on = 'id_cliente', how = 'inner')
#df_nome_tabela = pd.merge(tabela da esquerda, tabela da direita, left_on = 'nome do campo da tabela da esquerda, right_on= 'nome do campo da outra tabela)
display(df_catalago_produtos_vendas_pedidos_perfil_clientes)
```

```python
# MARGEM DE LUCRO BRUTO
df_catalago_produtos_vendas_pedidos_perfil_clientes['Margem de Lucro Bruto'] = (df_catalago_produtos_vendas_pedidos_perfil_clientes['preco_venda']-df_catalago_produtos_vendas_pedidos_perfil_clientes['custo_producao'])*df_catalago_produtos_vendas_pedidos_perfil_clientes['quantidade']

display(df_catalago_produtos_vendas_pedidos_perfil_clientes.head(50))
```

```python
# Calcular quantos itens há na coluna de quantidade e quantos há na coluna de preco_venda
# Realiza a soma de cada coluna
total_quantidade = df_catalago_produtos_vendas_pedidos_perfil_clientes['quantidade'].sum()
faturamento_total = df_catalago_produtos_vendas_pedidos_perfil_clientes['preco_venda'].sum()
# realiza o ticket médio de todas as vendas independente do produto
ticket_medio_venda = faturamento_total / total_quantidade

# Exibe os resultados
print(f"total_quantidade: {total_quantidade}")
print(f"faturamento_total R$: {faturamento_total:.2f}")
print(f"Ticket Médio de vendas: R$ {ticket_medio_venda:.2f}")
```

```python
# para saber quantos clientes únicos
total_clientes_unicos = df_catalago_produtos_vendas_pedidos_perfil_clientes['id_cliente'].nunique()

print(f"Total de clientes únicos: {total_clientes_unicos}")
```

```python
# para saber ticket médio POR Clientes
ticket_medio_por_cliente = faturamento_total / total_clientes_unicos

print(f"O Ticket Médio por cliente na base é: R$ {ticket_medio_por_cliente:.2f}")
```
